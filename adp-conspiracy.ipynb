```python
from pandas import read_csv
import seaborn
import matplotlib.pyplot as plt

espn = read_csv("espn.csv", index_col=False)
yahoo = read_csv("yahoo.csv")
yahoo["Drop"] = yahoo["ADP"] - yahoo["Rank"]
```

```python
min(yahoo["Drop"])
```

```python
import matplotlib.pyplot as plt
import numpy as np

color = "#4470AD"

fig, ax = plt.subplots()
counts, bins, patches = ax.hist(
    espn["Drop"],
    range=(-40, 40),
    bins=16,
    edgecolor="none",
    color=color,
)

# Add labels only for non-zero counts
ax.bar_label(
    patches,
    labels=[str(int(c)) if c > 0 else "" for c in counts],
    padding=2,
    color=color,
    weight="bold"
)

# Remove spines, y-axis ticks, and gridlines for clean look
for i, spine in enumerate(ax.spines.values()):

    spine.set_visible(False)
ax.yaxis.set_ticks([])
ax.grid(False)
ax.tick_params(axis='x', length=0)

# Add a vertical line at the middle of the x-axis
middle = (bins[0] + bins[-1]) / 2
ax.vlines(x=middle, ymin=0, ymax=max(counts), color='black', linewidth=1)

plt.show()
```
